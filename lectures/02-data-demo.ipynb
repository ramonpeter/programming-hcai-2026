### `=` is not equals

In mathematics `x = x + 1` has no solution. In Python it is an instruction:
*work out the right-hand side, then attach the result to the name on the left.*

**Read every assignment right to left.**

````python
x = 5
print(x)
````

````python
x = x + 1      # right side first: 5 + 1 = 6
print(x)
````

````python
x = x * 2      # 6 * 2 = 12
print(x)
````

## 3 · Operators and precedence

Same order as in mathematics: brackets, powers, multiply/divide, add/subtract.
Ties break **left to right**.

````python
print(2 + 3 * 4)
print((2 + 3) * 4)
````

````python
print(8 / 4 * 2)     # left to right — not 1.0
````

````python
print(2 ** 3 ** 2)   # ** is the exception: right to left
````

### Three kinds of division

````python
print(17 / 5)        # true division  — always a float
print(17 // 5)       # floor division — rounds DOWN
print(17 % 5)        # modulo         — the remainder
````

`//` and `%` together split a number into *how many fit* and *what is left*:

````python
seconds = 3725
print(seconds // 60, "minutes and", seconds % 60, "seconds")
````

And `%` is how you test divisibility:

````python
n = 12
print(n % 2 == 0)    # True means n is even
````

**Careful with negatives.** `//` rounds *down*, not towards zero:

````python
print(-7 // 2)       # -4, not -3
print(-7 % 2)
````

## 4 · Floats are approximations

The thing we promised to explain last week.

````python
print(0.1 + 0.2)
````

````python
print(0.1 + 0.2 == 0.3)
````

A float is stored in a fixed number of bits. One tenth in binary never terminates —
exactly like one third in decimal — so Python stores the closest value it can.

**Never compare two floats with `==`.** Compare the difference to a small tolerance:

````python
print(abs(0.1 + 0.2 - 0.3) < 1e-9)
````

Integers are exact, and Python integers have no size limit:

````python
print(2 ** 200)
````

## 5 · Changing type on purpose

````python
print(int("42") + 1)
print(float("3.14"))
print(str(42) + "!")
````

`int()` on a float **chops**, it does not round:

````python
print(int(3.9))
print(round(3.9))
````

And when the contents are wrong, you get a `ValueError`:

````python
int("3.14")
````

The type was right — it was a `str`, as `int()` expects. The *contents* were unusable.
That is the difference between `ValueError` and `TypeError`.

| error | means |
|---|---|
| `NameError` | there is no such name |
| `TypeError` | wrong kind of value |
| `ValueError` | right kind, bad contents |
