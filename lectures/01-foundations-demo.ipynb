### Notebook vs script

Notice that the cells above printed their result **without** `print()`. That is a notebook
convenience: it shows the value of the last line of a cell.

A `.py` script does **not** do this. In a script you must say `print()` explicitly, or you
see nothing. This surprises everybody exactly once — usually at 2 a.m.

Compare:

````python
7 + 3
2 + 2         # only this one is shown — the last line wins
````

````python
print(7 + 3)
print(2 + 2)  # both are shown, and this works in a script too
````

### 4.1 — SyntaxError

````python
print("Hello, world)
````

The closing quote is missing, so Python never finds where the text ends.

`SyntaxError` means Python could not even *read* your program. Nothing ran at all.

Look at the reported line **and the line above it** — an unclosed bracket or quote is often
reported one line late.

### 4.2 — IndentationError

````python
print("first")
  print("second")
````

In Python, leading spaces are not decoration — they are part of the grammar. A line that
starts further right means *this belongs inside the thing above*. Nothing is above it here,
so the indent makes no sense.

From lecture 03 onwards, indentation is how you say what is inside an `if` or a loop.

### 4.3 — NameError

````python
message = "Hello"
print(mesage)
````

Python knows the names you gave it, and nothing else. One letter wrong is a different
name, and Python has no idea what you meant. 

Run the same code in a script or in the REPL and Python 3.10+ adds
`Did you mean: 'message'?`. Notebooks do not show it, because Jupyter formats
tracebacks itself. Same error, less help.

Names are also **case-sensitive**: `Message` and `message` are two different things.

### 4.4 — TypeError

````python
print("Result: " + 42)
````

`+` means *join* for text and *add* for numbers. Python refuses to mix the two.

There are two fixes. Both are correct:

````python
print("Result: ", 42)          # a comma instead of a plus
````

````python
print("Result: " + str(42))    # convert the number to text first
````

## 5 · Reading an error message

```
Traceback (most recent call last):
  File "hello.py", line 3, in <module>
    print(greting)
NameError: name 'greting' is not defined
```

1. Read the **last line first** — it names the problem.
2. Then the **line number** — it says where.
3. The middle is the path Python took to get there. Ignore it for now.

### When you are stuck

| | |
|---|---|
| 1 | Read the last line of the error |
| 2 | Go to the line number — and look at the line above it too |
| 3 | Put a `print()` before the failing line and look at what the values actually are |
| 4 | Change **one** thing, then run again. Never three at once |
| 5 | Ask — a classmate, then me. Bring the error message, not "it does not work" |

Nobody writes code without errors. Experienced programmers just read them faster.
