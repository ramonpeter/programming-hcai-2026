## 2 · input() always gives you text

This is the bug of the week. You type digits, but what comes back is a **str**.

```python
age = input("Your age: ")
print(type(age))
```

```python
print(age + 1)
```

`TypeError`. Even though you typed `24`, `age` is the text `"24"`.

Read the fix from the **inside out**: `input()` runs first, then `int()` wraps it.

```python
age = int(input("Your age: "))
print(age + 1)
```

Use `float()` when the answer can have decimals:

```python
price = float(input("Price: "))
print(f"With VAT: {price * 1.22:.2f}")
```

And use **neither** when the value is genuinely text. A name is text. So is a
postcode — `int()` would eat its leading zero.

**Try it:** ask for two numbers and print their sum. Get it wrong first (no `int()`),
look at what it prints, then fix it.

## 4 · f-strings

The one piece of syntax worth memorising today.

```python
name = "Mirea"
marks = 28

print(f"{name} scored {marks} out of 30")
```

- the `f` goes **before** the opening quote;
- anything inside `{ }` is evaluated;
- the value is converted to text for you — no more `str()`.

Anything Python can compute goes inside the braces:

```python
print(f"{marks} out of 30 is {marks / 30 * 100:.1f}%")
```

### Formatting: everything after the colon

Predict each one before running it. The `|` is there so you can count the spaces.

```python
pi = 3.14159
print(f"{pi:.2f}")
print(f"{pi:.0f}")
```

```python
print(f"{42:>8}|")      # right-aligned in 8 columns
print(f"{'ab':<8}|")     # left-aligned
print(f"{'ab':^8}|")     # centred
```

```python
print(f"{1234567:,}")
```

A width is a **minimum**: it pads, it never truncates. Numbers default to the right,
text to the left — the arrow only matters when you want the other one.

You do not need to memorise these. You need to know they exist.

## 5 · Worked example — the receipt that would not line up

Sheet 1 asked for this table and we spaced it by hand. It was fragile: change a name and
the columns break. Let us do it properly.

Start with no formatting at all:

```python
name, qty, price = "Coffee", 3, 1.50
print(f"{name}{qty} x {price} = {qty * price}")
```

Unreadable. Now add widths, one at a time:

```python
print(f"{name:<10}{qty:>3} x {price:>6} = {qty * price:>7}")
```

Columns, but the prices are not prices. Add `.2f`:

```python
print(f"{name:<10}{qty:>3} x {price:>6.2f} = {qty * price:>7.2f}")
```

Now the whole receipt:

```python
print(f"{'Coffee':<10}{3:>3} x {1.50:>6.2f} = {3 * 1.50:>7.2f}")
print(f"{'Cornetto':<10}{2:>3} x {1.15:>6.2f} = {2 * 1.15:>7.2f}")
print(f"{'Spremuta':<10}{1:>3} x {3.00:>6.2f} = {1 * 3.00:>7.2f}")
print(f"{'Total':<10}{'':>3}   {'':>6}   {3*1.50 + 2*1.15 + 3.00:>7.2f}")
```

## 6 · Conditions

A condition is a question with a **yes/no** answer. Python calls those answers
`True` and `False`.

```python
print(7 > 3)
print(7 == 3)
print(7 != 3)
print("Yes" == "yes")
```

Note the last one: comparing text is **case-sensitive**.

And note `==`, with two equals signs.

| | |
|---|---|
| `=` | **assigns** a value |
| `==` | **asks** whether two things are equal |

### Combining conditions

```python
age = 20
has_ticket = True

print(age >= 18 and has_ticket)
print(age >= 18 or has_ticket)
print(not has_ticket)
```

**English is looser than Python.** "if the mark is 28 or 30" has to be written out in
full — each comparison needs its own left-hand side. Here is what happens if you do not:

```python
mark = 12
print(mark == 28 or 30)
```

It printed `30` — not `False`, and not even `True`.

Python read it as `(mark == 28) or (30)`. The first part is `False`, so `or` hands back the
second one, which is the bare number `30`. And in a condition, any non-zero number counts
as True:

```python
print(28 or 30)
print(bool(30))
```

So `if mark == 28 or 30:` runs its branch for *every* mark.

This is a **silent** bug: no error, wrong answer. The correct version:

```python
print(mark == 28 or mark == 30)
```

## 7 · if, elif, else

```python
mark = 24

if mark >= 28:
    print("excellent")
elif mark >= 24:
    print("good")
else:
    print("pass")
```

- the colon ends the question;
- the indented block is what happens if it holds;
- `elif` is checked **only** if everything above it failed;
- **exactly one** branch runs.

### Indentation is the grammar

Same words, different program. Run both with `mark = 10`:

```python
mark = 10

if mark >= 18:
    print("pass")
    print("well done")
```

```python
mark = 10

if mark >= 18:
    print("pass")
print("well done")
```

The second one congratulates a failing student. The only difference is four spaces.

## 8 · Worked example — a grade classifier, written twice

First the way most people write it:

```python
mark = 26

if mark >= 18:
    if mark >= 24:
        if mark >= 28:
            print("excellent")
        else:
            print("good")
    else:
        print("pass")
else:
    print("fail")
```

Correct, and hard to read — every band adds a level of indentation.

Now flat, with `elif`:

```python
mark = 26

if mark >= 28:
    print("excellent")
elif mark >= 24:
    print("good")
elif mark >= 18:
    print("pass")
else:
    print("fail")
```

Same behaviour. Every band is one line, and they read in order.

**When you find yourself indenting a third time, ask whether `elif` would flatten it.**

### Order matters

The discount rule: 10% over €50, 20% over €100.

```python
amount = 150

if amount > 50:
    discount = 0.10
elif amount > 100:
    discount = 0.20
else:
    discount = 0

print(discount)
```

10% on €150. Anything over 100 is *also* over 50, so the first branch catches it and
the 20% band can never run.

No error. It just charges the wrong price. **Most specific first:**

```python
amount = 150

if amount > 100:
    discount = 0.20
elif amount > 50:
    discount = 0.10
else:
    discount = 0

print(discount)
```

## 10 · Worked example — leap year

You wrote this algorithm in English in **sheet 1, A1**. Here it is as a program.

```python
year = int(input("Year: "))

if year % 4 != 0:
    leap = False
elif year % 100 != 0:
    leap = True
elif year % 400 != 0:
    leap = False
else:
    leap = True

print(f"{year}: leap year = {leap}")
```

Test it on **1900** (not a leap year) and **2000** (a leap year). If it gets 1900
right, the ordering is correct.

Notice what the program does *not* do: it never prints inside a branch. Each branch sets
one variable, and the printing happens once, at the end. That habit scales; printing in
every branch does not.

**Try it:** what happens if you swap the second and third `elif`? Predict first, then try
