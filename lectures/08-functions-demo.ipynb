## A · From a specification, and from old code

### A1 · days_alive

The specification is one sentence: *how many days has someone of a given age been
alive?* Before any code, three decisions.

- **What goes in?** An age in whole years. One parameter, `age`.
- **What comes back?** A number of days, a whole number.
- **What should it give?** Three cases worked out by hand, to test against.

```python
# days_alive(0)  should be 0
# days_alive(1)  should be 365
# days_alive(20) should be 7300
```

A first attempt.

```python
def days_alive(age):
    """Return roughly how many days someone aged age years has lived."""
    print(age * 365)

days_alive(20)
```

It shows 7300, which looks right. Now use it the way a test does.

```python
print(days_alive(20) == 7300)
```

7300 appears, and then `False`. The function *showed* the number and handed back `None`,
and `None == 7300` is false. That is return versus print from the slides, caught by the
first test. Fix the one word.

```python
def days_alive(age):
    """Return roughly how many days someone aged age years has lived."""
    return age * 365

print(days_alive(0), days_alive(1), days_alive(20))
print(days_alive(20) == 7300)
```

All three tests pass. Now the specification gets sharper: *count the leap days, roughly
one every four years.* The tests change first, then the body.

```python
def days_alive(age):
    """Return roughly how many days someone aged age years has lived.

    Counts one leap day for every four years.
    """
    return age * 365 + age // 4

print(days_alive(0), days_alive(1), days_alive(20))      # 0 365 7305
```

Still "roughly": the true answer depends on the birthday, which the specification does
not give us. A docstring that says *roughly* is honest about that.

### A2 · The exam session, rewritten

This is the program from lecture 6, written as one block. It works.

```python
grades = [28, 30, 24, 15, 27, 18, 12, 29]

print(f"mean:    {sum(grades) / len(grades):.1f}")
passed = []
for g in grades:
    if g >= 18:
        passed.append(g)
print(f"passed:  {len(passed)} of {len(grades)}")
print(f"ranking: {sorted(passed, reverse=True)}")
failed, low, middle, high = 0, 0, 0, 0
for g in grades:
    if g < 18:
        failed += 1
    elif g <= 23:
        low += 1
    elif g <= 27:
        middle += 1
    else:
        high += 1
print(f"bands:   {failed} {low} {middle} {high}")
```

Every chunk of it answers one question. Give each question a function, with a name, a
docstring and a test. Start with the smallest.

```python
def mean(xs):
    """Return the mean of the numbers in xs. xs must not be empty."""
    return sum(xs) / len(xs)

print(mean([30, 18]))         # 24.0
```

```python
def passes(grades):
    """Return a new list of the grades that pass, 18 or more, in the same order."""
    result = []
    for g in grades:
        if g >= 18:
            result.append(g)
    return result

print(passes([15, 18, 30]))   # [18, 30]
```

A function can call another one. `ranking` does not repeat the pass rule, it asks `passes`.

```python
def ranking(grades):
    """Return the passing grades, best first."""
    return sorted(passes(grades), reverse=True)

print(ranking([28, 15, 30]))  # [30, 28]
```

Four answers come back as one tuple, as on the slide about `min_max`.

```python
def bands(grades):
    """Return how many grades are failed, 18-23, 24-27 and 28-30, as a tuple."""
    failed, low, middle, high = 0, 0, 0, 0
    for g in grades:
        if g < 18:
            failed += 1
        elif g <= 23:
            low += 1
        elif g <= 27:
            middle += 1
        else:
            high += 1
    return failed, low, middle, high

print(bands([15, 20, 25, 29]))   # (1, 1, 1, 1)
```

The printing goes in one function whose whole job is to print. That is the case where a
function that prints and returns nothing is right.

```python
def report(grades):
    """Print a summary of one exam session."""
    print(f"mean:    {mean(grades):.1f}")
    print(f"passed:  {len(passes(grades))} of {len(grades)}")
    print(f"ranking: {ranking(grades)}")
    failed, low, middle, high = bands(grades)
    print(f"bands:   {failed} {low} {middle} {high}")


def main():
    grades = [28, 30, 24, 15, 27, 18, 12, 29]
    report(grades)


main()
```

The same output as the block above. Three things changed. The pass mark 18 now lives in
one place, inside `passes`. Each piece was tested before it was used. And `main` reads
like the plan.

## B · A small calculator

**Specification.** Show a menu of four operations and `q` to quit. Read a choice and two
numbers. Print the result. Repeat until `q`.

### Step 1 · main first

Written before any of the functions it calls exist. It cannot run yet, and that is fine:
it is the plan, in Python.

```python
def main():
    """Run the calculator until the user chooses q."""
    while True:
        op = read_choice()
        if op == "q":
            break
        a = read_number("first number:  ")
        b = read_number("second number: ")
        result = calculate(op, a, b)
        if result is None:
            print("cannot divide by zero")
        else:
            print("result:", result)
```

`main` needs three functions: `read_choice`, `read_number` and `calculate`. And
`calculate` will need one function per operation. Fill them in from the bottom.

### Step 2 · One function per operation

```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    """Return a / b, or None when b is 0."""
    if b == 0:
        return None
    return a / b

print(add(2, 3), subtract(2, 3), multiply(2, 3), divide(1, 4), divide(1, 0))
```

`divide` decides nothing about what the user sees. It returns `None`, and `main` decides
what to print. Functions that compute return, the function that talks to the user prints.

### Step 3 · Choosing the operation

```python
def calculate(op, a, b):
    """Apply the operation op, one of + - * /, to a and b."""
    if op == "+":
        return add(a, b)
    elif op == "-":
        return subtract(a, b)
    elif op == "*":
        return multiply(a, b)
    else:
        return divide(a, b)

print(calculate("+", 2, 3), calculate("/", 6, 4))
```

### Step 4 · Reading input

`float()` rather than `int()`, so 2.5 works. What happens on `twenty` is lecture 9.

```python
def read_number(prompt):
    """Ask once with prompt, and return the answer as a float."""
    return float(input(prompt))
```

The menu. A first attempt at checking the choice.

```python
def read_choice():
    """Show the menu, and return a valid choice: + - * / or q."""
    print("+  add    -  subtract    *  multiply    /  divide    q  quit")
    op = input("choice: ").strip()
    while op not in "+-*/q":
        op = input("choice: ").strip()
    return op

print("" in "+-*/q", "+-" in "+-*/q")
```

Both `True`. `in` on a string asks *is this a piece of the string?*, so pressing Enter on
an empty line, or typing `+-`, counts as valid. We want *is it one of these five
things?*, which is the question a set answers.

```python
VALID = {"+", "-", "*", "/", "q"}

def read_choice():
    """Show the menu, and return a valid choice: + - * / or q."""
    print("+  add    -  subtract    *  multiply    /  divide    q  quit")
    op = input("choice: ").strip()
    while op not in VALID:
        print("please choose one of + - * / q")
        op = input("choice: ").strip()
    return op

print("" in VALID, "+-" in VALID, "+" in VALID)
```

### Step 5 · Run it

Try a wrong choice, a sum, and a division by zero. Then `q`.

```python
main()
```

### Step 6 · A dictionary of functions

`calculate` is a chain of `elif`s that maps a symbol to a function. Mapping keys to values
is what a dictionary does, and functions are values. The whole chain becomes one lookup.

```python
OPERATIONS = {"+": add, "-": subtract, "*": multiply, "/": divide}

def calculate(op, a, b):
    """Apply the operation op, one of the keys of OPERATIONS, to a and b."""
    return OPERATIONS[op](a, b)

print(calculate("*", 6, 7), calculate("/", 1, 0))
```

Read `OPERATIONS[op](a, b)` left to right. `OPERATIONS[op]` is a function, for example
`multiply`, and `(a, b)` calls it. No brackets inside the dictionary: we store the
functions, we do not call them. `main` did not change at all, because it only ever
asked `calculate`.
