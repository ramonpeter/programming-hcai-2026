## 2 · while

Four parts. Learn to name them — you will be looking for the missing one all semester.

```
i = 1                 # 1. set up
while i <= 5:         # 2. the condition
    print(i)          # 3. the body
    i = i + 1         # 4. the change
```

````python
i = 1

while i <= 5:
    print(i)
    i = i + 1

print("done, i is now", i)
````

Five passes, and then `i` is 6 and the condition fails.

Notice the last line: the condition was checked a **sixth** time. A loop always tests once
more than it repeats.

## 4 · The accumulator

A variable that lives **outside** the loop and grows **inside** it.

- starts at the neutral value: `0` for a sum, `1` for a product;
- counting is the same pattern: `count = count + 1`;
- `total += x` is the shorthand from lecture 2.

**Declared outside, updated inside, read after.**

Here is what happens when you get that shape wrong:

````python
i = 1
while i <= 5:
    total = 0                # the bug: this wipes the total every pass
    total = total + i
    i = i + 1

print(total)
````

`5`, not `15`. Each pass computed `0 + i`, and what survived was the last one.

Note *why* this is hard to spot: the answer is plausible. For `n = 1` the program is even
correct. Fix it by moving one line:

````python
total = 0                    # outside
i = 1
while i <= 5:
    total = total + i        # inside
    i = i + 1

print(total)                 # after
````

## 5 · Three ways a loop goes wrong

**1 — it never starts.** The condition is False at the very first check.

````python
i = 10
while i < 5:
    print("you will never see this")
    i = i + 1

print("the loop body never ran, and nothing complained")
````

**2 — it never stops.** Nothing in the body moves towards making the condition False.

The cell below has a **safety net** so this notebook does not hang. The bug is real; the
`break` is not part of it.

````python
x = 10
steps = 0

while x > 0:
    x = x + 1          # the bug: should be x - 1
    steps = steps + 1
    if steps > 20:     # safety net - not part of the example
        print("still going after 20 passes. This loop never ends.")
        break
````

This is **sheet 1, A5** — the loop you fixed on paper in week one.

### When it really does hang

Take the safety net out and run it, on purpose, once:

- **in a notebook:** the square ■ Stop button, or *Kernel → Interrupt*
- **in a terminal:** `Ctrl + C`

The program stops where it was. Nothing is broken. A cell showing `[*]` for ever is not
frozen — it is doing exactly what you told it to, very fast, for ever.

Do this once deliberately today. The first accidental one is much less alarming afterwards.

**3 — it runs the wrong number of times.** Almost always at the end, and almost always
by one.

````python
# "the numbers 1 to 5"
i = 1
while i < 5:        # < or <= ?
    print(i)
    i = i + 1
````

## 6 · Worked example — numbers until someone says stop

The plan first:

```
# a running total, and a count
# ask for a number
# while it is not 0:
#     add it in, count it
#     ask again
# print the total and the average
```

````python
total = 0
count = 0

value = int(input("Number (0 to stop): "))

while value != 0:
    total += value
    count += 1
    value = int(input("Number (0 to stop): "))

print(f"Total: {total}")
print(f"Average: {total / count:.2f}")
````

Two things to notice.

**The question is asked twice** — once before the loop, once at the end of the body. That
is the shape of every loop that reads until a stopping value.

**What happens if the very first answer is 0?** `count` stays 0, and the average divides by
zero. Try it. Then decide what the program *should* do.

## 7 · break and continue

````python
total = 0

while True:
    value = int(input("Number (0 to stop): "))
    if value == 0:
        break
    total += value

print("Total:", total)
````

`break` leaves the loop immediately. No duplicated question — the exit sits in the
middle, which is exactly where it belongs.

`while True:` with a `break` is a normal, readable loop. What it is not is an excuse to
stop thinking about why the loop ends.

`continue` skips to the next pass:

````python
i = 0
while i <= 10:
    i += 1                 # the change comes FIRST
    if i % 2 != 0:
        continue           # skip the odd ones
    print(i)
````

**Careful:** if the change came *after* the `continue`, the loop would skip it and hang
for ever. Use `continue` sparingly — an `if` around the body usually reads better.

## 8 · for and range

When you know how many times, `for` does steps 1 and 4 for you.

````python
for i in range(5):
    print(i)
````

The counter is created for you, and it cannot run away.

It starts at **0** and stops **before** 5 — five numbers, none of them 5.

````python
print(list(range(5)))
print(list(range(2, 7)))
print(list(range(0, 10, 3)))
print(list(range(5, 0, -1)))
````

````python
print(list(range(3, 3)))      # nothing
print(list(range(10, 0)))     # nothing - needs a negative step
````

`list()` is just so we can see them all at once; you will not normally write it.

**The stop value is never included.** So the numbers 1 to n, *including* n:

````python
n = 5
for i in range(1, n + 1):
    print(i, end=" ")
````

## 9 · while or for?

One question: **do I know, before the loop starts, how many times it will run?**

| Yes → `for` | No → `while` |
|---|---|
| five times, n times | until the user types something |
| every number in a range | until the answer is good enough |
| every item in something (lecture 6) | until a condition changes |

Anything a `for` loop can do, a `while` loop can do — with more code and more ways to get
it wrong. Reach for `for` first.

## 10 · Worked example — a times table

```
# for each row 1..5
#     for each column 1..5
#         print the product, padded
#     end the line
```

Build it in three steps. First the inner loop alone:

````python
for col in range(1, 6):
    print(f"{3 * col:>4}", end="")
````

Now wrap it in the outer loop:

````python
for row in range(1, 6):
    for col in range(1, 6):
        print(f"{row * col:>4}", end="")
    print()
````

The `end=""` keeps a row on one line — lecture 3 earning its keep. The bare `print()`
at the **outer** level ends the row.

**Try it:** comment out the bare `print()` and run it again. Why does everything collapse
onto one line?

## 11 · Nested loops

````python
for i in range(3):
    for j in range(4):
        print(i, j, end="   ")
    print()
````

- the outer loop runs 3 times;
- for each of those, the inner runs 4 times;
- so the body runs 3 × 4 = **12** times;
- the inner counter goes all the way round before the outer one moves.

Use different names for the two counters. `i` and `j` is the convention; `i` and `i` is a
bug you will spend an hour on.

Cost adds up fast: two nested loops over n items do n² passes. For n = 1000 that is a
million.

## 12 · Worked example — is it a prime?

````python
n = int(input("A number: "))

is_prime = True

for d in range(2, n):
    if n % d == 0:
        is_prime = False
        break

if is_prime and n > 1:
    print(f"{n} is prime")
else:
    print(f"{n} is not prime")
````

A **flag**: one bool that remembers what the loop found.

- set it to `True` and let the loop disprove it;
- `break` the moment you know — there is nothing left to learn;
- `n % d == 0` is the divisibility test from lecture 2;
- the `n > 1` guard catches 1 and 0, which are not prime.

Why `range(2, n)` and not `range(2, n + 1)`? Because n divides itself, and that would make
every number non-prime.

The flag-plus-break shape answers every *"is there any…?"* question you will ever write.
