## 1 · A string is a sequence of characters

Each character has a number: its **index**. Counting starts at 0.

```
 P   y   t   h   o   n
 0   1   2   3   4   5
-6  -5  -4  -3  -2  -1
```

````python
s = "Python"

print(s[0])
print(s[3])
print(len(s))
````

````python
print(s[-1])      # the last one
print(s[-2])      # the one before it
````

Negative numbers count back from the end. That saves you writing `s[len(s) - 1]`.

Go one past the end and Python objects:

````python
s[6]
````

`IndexError`. The last valid index is `len(s) - 1` — the same off-by-one as last week.

## 2 · Slicing

`s[start:stop]` — and **stop before**, exactly like `range()`.

````python
p = "Programming"

print(p[0:4])
print(p[4:7])
````

````python
print(p[:4])      # leave out the start: from the beginning
print(p[4:])      # leave out the end: to the end
print(p[:])       # the whole thing
print(p[-4:])     # the last four
````

**Try it:** how many characters does `p[2:9]` give? Predict, then run it.

### With a step

`s[start:stop:step]` — the same three numbers as `range()`.

````python
print(p[0:8:2])   # every second, from 0 up to 8
print(p[::2])     # every second, whole string
print(p[::-1])    # backwards
````

`s[::-1]` reverses a string. You will meet it constantly — worth recognising on sight.

There is no `reverse()` for strings, and there does not need to be.

And a slice never raises an error, however silly the numbers:

````python
print(repr(p[50:90]))    # "" - nothing
print(repr(p[5:2]))      # "" - also nothing
print(repr(p[4:4]))      # "" - start and stop equal
````

`repr()` shows the quotes, so you can see that the result really is an empty string.

Convenient — and occasionally dangerous. A slice that silently returns nothing is a bug
that does not announce itself.

## 4 · Strings are immutable

You can read a character out. You cannot write one back in.

````python
word = "cat"
word[0] = "b"
````

`TypeError: 'str' object does not support item assignment`.

What you do instead is build a **new** string:

````python
word = "cat"
word = "b" + word[1:]
print(word)
````

The string `"cat"` never changed. A new string was built, and the name `word` was
re-attached to it.

This is not a limitation to work around. It is why you can pass a string anywhere without
worrying that something will alter it behind your back.

**Lists, next week, are the opposite** — and that one difference causes all the trouble.

## 5 · Looping over text

A string is something you can iterate, so `for` works straight away.

````python
for ch in "Python":
    print(ch)
````

The accumulator from last week, walking over characters instead of numbers:

````python
text = "programming"
vowels = 0

for ch in text:
    if ch in "aeiou":
        vowels += 1

print(vowels)
````

`ch in "aeiou"` is worth a second look. `in` asks whether one string appears inside
another, and a single character *is* a string — so membership in a five-character string is
a compact way to say "is it a vowel".

**Try it:** count the digits in `"a1b2c3"` the same way.

## 6 · Escape sequences

Characters you cannot simply type.

````python
print("first\nsecond")      # \n is a new line
print("a\tb\tc")            # \t is a tab
````

````python
print("She said \"hello\" and left")
print('She said "hello" and left')     # easier: swap the quotes
````

**Sheet 1, B3 number 6** was exactly this. The answer then was "swap the quotes", and it
still is. An escape is what you reach for when both kinds of quote appear in one string.

One to remember for lecture 9: a backslash starts an escape, so a Windows path like
`"C:\new"` contains a newline, not the letter n.

## 7 · Worked example — is it a palindrome?

### Pass 1 — naive

````python
text = "level"

if text == text[::-1]:
    print("palindrome")
else:
    print("not a palindrome")
````

Works. Now try it with a capital:

````python
text = "Level"

if text == text[::-1]:
    print("palindrome")
else:
    print("not a palindrome")
````

### Pass 2 — ignore capitals

````python
text = "Level"
text = text.lower()

if text == text[::-1]:
    print("palindrome")
else:
    print("not a palindrome")
````

### Pass 3 — ignore everything that is not a letter

````python
text = "A man, a plan, a canal: Panama"

clean = ""
for ch in text.lower():
    if ch in "abcdefghijklmnopqrstuvwxyz":
        clean = clean + ch

print(clean)
print(clean[::-1])
print(clean == clean[::-1])
````

Pass 3 is the accumulator again — building a **string** instead of a number.

Naive, then robust. That progression is the lesson; the finished version on its own teaches
much less.

## 8 · A new piece of syntax: the dot

````python
text = "  Hello World  "

print(repr(text.lower()))
print(repr(text.strip()))
print(repr(text.upper()))
````

- `len(s)` is a **function**: the string goes inside the brackets.
- `s.lower()` is a **method**: the string comes first, then a dot, then the name.
- A method belongs to the thing it is attached to. Only strings have `.lower()`.
- The brackets are still there, and still mean *run it now*.

The dot is how Python says "belonging to". You will see it everywhere from here on — and in
lecture 11 you will write your own.

## 9 · The ones worth knowing

````python
t = "  Ada  Lovelace  "

print(repr(t.strip()))
print(repr(t.replace(" ", "")))
print(t.count("a"))
print(t.find("Lov"))
print(t.strip().startswith("Ada"))
````

Note `t.count("a")` gives **2**, not 3. The capital `A` does not match — everything about
strings is case-sensitive.

You are not expected to memorise the list. You are expected to know that string methods

## 10 · The bug this causes

````python
name = "ada"
name.upper()
print(name)
````

`ada`. Nothing happened — and nothing complained.

`.upper()` handed back a new string, and because we did not catch it, it was thrown away.

````python
name = "ada"
name = name.upper()
print(name)
````

**A string method never changes the string.** It returns a new one, and if you do not
assign it to something, it is gone.

This is the single most common string bug, and it is completely silent. Whenever a string
operation "does not work", check first whether you caught the result.

## 11 · split and join

````python
sentence = "the quick brown fox"

words = sentence.split()
print(words)
print(len(words))
````

````python
for w in words:
    print(w)
````

What comes back is a **list** — the square brackets are the clue. That is next week.
For today: you can loop over it and ask for its `len()`.

`split(",")` splits on commas instead of spaces — useful for data files in lecture 9.

And `join` is `split` backwards. The string you call it on is the glue:

````python
print("-".join(words))
print(" ".join(words))
````

## 12 · Worked example — statistics on a paragraph

````python
text = "the quick brown fox jumps over the lazy dog"

print(f"characters: {len(text)}")
print(f"words:      {len(text.split())}")
````

````python
longest = ""

for w in text.split():
    if len(w) > len(longest):
        longest = w

print("longest:", longest)
````

That loop is the **running maximum** from sheet 4, B7 — with `len(w)` in place of the
number itself. Same shape, third outing: largest of three (lecture 1), largest of n
(sheet 4), longest word (today).

### Letter frequency

````python
low = text.lower()

for letter in "abcdefghijklmnopqrstuvwxyz":
    n = low.count(letter)
    if n > 0:
        print(letter, n)
````

A loop over the alphabet, asking the text how many of each it holds.

Notice what this costs: it reads the whole text twenty-six times, once per `count()`. For a
paragraph nobody cares. For a book it is slow. Lecture 7 has the structure that does it in
one pass.
