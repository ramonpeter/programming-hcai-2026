## A · Who has not handed in?

Eight students are on the class list. Thirteen submissions came in, and each one carries
the name the student typed.

```python
class_list = ["Ada", "Alan", "Barbara", "Grace", "Katherine", "Linus", "Margaret", "Tim"]

submissions = ["Grace", "alan", "Ada", "Grace", " Margaret", "Tim", "linus ",
               "Ada", "Barbra", "Grace", "alan", "Tim", "Margaret"]
```

We want three answers. Write them down as a plan before any code.

```python
# 1. how many different students handed in?
# 2. who on the class list has not handed in?
# 3. which submitted names are not on the class list at all?
```

### Step 1 · How many handed in?

Thirteen submissions, but some students submitted twice. A set keeps each value once.

```python
print(len(submissions))
print(len(set(submissions)))
```

Eight different names, and eight students on the list. Everyone handed in?

Look at what is actually in the set before believing that.

```python
print(set(submissions))
```

`" Margaret"` and `"Margaret"` are different strings, so the set keeps both. The 8 is a
coincidence. Check it from the other side.

### Step 2 · Who is missing? A first attempt

```python
missing = set(class_list) - set(submissions)
print(missing)
```

Alan and Linus handed in. Python is right, though: `"alan"` is not `"Alan"`, and
`"linus "` has a space at the end. The data needs cleaning before any comparison means
anything.

### Step 3 · Clean the names first

`strip()` removes the spaces at the ends and `title()` gives one capital letter at the
start. Both are from lecture 5. Collect the clean names in a set as we go.

```python
submitted = set()
for name in submissions:
    submitted.add(name.strip().title())

print(len(submitted))
print(sorted(submitted))
```

Printing `sorted(...)` instead of the set itself gives a list in alphabetical order. A set
has no order of its own, so printing it directly can come out differently next time.

### Step 4 · The three answers

```python
everyone = set(class_list)

missing = everyone - submitted
unknown = submitted - everyone

print("handed in:", len(submitted & everyone), "of", len(everyone))
print("missing:  ", sorted(missing))
print("unknown:  ", sorted(unknown))
```

`Barbra` is not on the list, and `Barbara` has not handed in. Almost certainly this is the
same person with a typo. The program cannot know that, and should not guess. It found the
two names that need a human to look at them, which is the useful part.

So the honest result is six handed in, Katherine is missing, and Barbara needs checking.

**Try it:** which students submitted more than once? A set throws exactly that
information away. Come back to this after part B, which has the structure that keeps it.

## B · Which words does a text use most?

Lecture 5 counted letters by asking the text 26 times, one `count()` per letter. Here we
count words, and we read the text once.

```python
text = """The Analytical Engine was a machine that existed only on paper. Ada Lovelace
wrote the notes that made it famous. In the notes she described how the engine
could work with symbols as well as numbers, and she wrote out a method for the
engine to compute a sequence of numbers. The machine was never built, but the
notes were read again a century later, when the first electronic machines were
built. Many people now call the method in her notes the first program."""
```

```python
# 1. split the text into words
# 2. count each word in one pass
# 3. how many words, how many different words
# 4. the five most frequent words
# 5. the same, without the small words
```

### Step 1 · Split and count

The counting pattern from the slides: how many so far, or 0 the first time, plus one.

```python
words = text.split()

counts = {}
for w in words:
    counts[w] = counts.get(w, 0) + 1

print(len(words), "words,", len(counts), "different")
```

Look at a few entries before trusting the numbers.

```python
print(counts["the"], counts["The"])
print(counts["numbers,"], counts["numbers."])
```

`"the"` and `"The"` are counted separately, and so are `"numbers,"` and `"numbers."`. To
Python they are different strings. Same lesson as part A: clean first, then count.

### Step 2 · Clean each word

`lower()` deals with the capitals. For the punctuation, `strip()` takes an argument: the
characters to remove from both ends. `"numbers.".strip(".,")` gives `"numbers"`.

```python
counts = {}
for w in words:
    w = w.lower().strip(".,;:!?")
    counts[w] = counts.get(w, 0) + 1

print(len(words), "words,", len(counts), "different")
print(counts["the"], counts["numbers"])
```

### Step 3 · The most frequent word

The running maximum from lecture 4, over the pairs that `.items()` hands us.

```python
best_word, best_count = "", 0
for w, n in counts.items():
    if n > best_count:
        best_word, best_count = w, n

print(best_word, best_count)
```

### Step 4 · The top five

For more than one we sort the words by their count. `key=counts.get` tells `sorted` to
compare words by the value `counts.get(word)`, not alphabetically. Use it as a recipe for
now. Lecture 8 explains why it works, together with `key=str.lower` from lecture 6.

```python
top = sorted(counts, key=counts.get, reverse=True)[:5]

for w in top:
    print(f"{w:<10} {counts[w]}")
```

Correct, and useless. The most frequent words in almost any English text are *the*, *a*,
*of*. They say nothing about what the text is about.

### Step 5 · Without the small words

Put the words to ignore in a set, because the only question we ask it is *is this word in
there?* That is the membership test from part A, now inside the counting loop.

```python
small = {"the", "a", "an", "of", "and", "in", "to", "was", "were", "as", "it",
         "that", "she", "her", "for", "with", "but", "how", "on", "out", "when",
         "could", "now", "only", "again", "well"}

counts = {}
for w in words:
    w = w.lower().strip(".,;:!?")
    if w not in small:
        counts[w] = counts.get(w, 0) + 1

top = sorted(counts, key=counts.get, reverse=True)[:5]
for w in top:
    print(f"{w:<10} {counts[w]}")
```

Now the top five say what the paragraph is about. Notice that *machine* and *machines*
are still counted separately. Fixing that properly is a hard problem, and it is where
natural language processing starts.

All three structures from today appeared in one program: a list of words, a dictionary of
counts, and a set of words to skip.
